## FOR LOOP

A `for` loop is used to repeatedly execute a block of code for each item in a sequence or iterable.

It can be used with:
- Strings
- Lists
- Tuples
- Sets
- Dictionaries
- `range()`

## SYNTAX

for variable in sequence:
    statement

## FOR LOOP WITH RANGE()

`range()` generates a sequence of numbers.

### SYNTAX

range(start, stop, step)

- `start` - starting value (default is 0)
- `stop` - ending value (excluded)
- `step` - increment/decrement value (default is 1)

## IMPORTANT POINTS

- The loop variable takes one value at a time from the sequence.
- The loop continues until all elements are processed.
- Indentation is mandatory.
- `break` can be used to terminate the loop.
- `continue` can be used to skip the current iteration.

## BASIC PROBLEMS

- Print numbers from 1 to 10
- Print even numbers from 1 to 20
- Print odd numbers from 1 to 20
- Print multiplication table
- Find the sum of numbers
- Find the factorial of a number
- Count vowels in a string
- Print elements of a list
- Find the largest number in a list

```python
# Print numbers from 1 to 10

for i in range(1, 11):
    print(i,end=" ")
```

```python
# Print even numbers from 1 to 20

for i in range(1, 21):
    if i % 2 == 0:
        print(i,end=" ")
```

```python
# Multiplication table

num = int(input("Enter a number: "))

for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

```python
# Find sum of numbers from 1 to n

n = int(input("Enter n: "))
total = 0

for i in range(1, n + 1):
    total += i

print("Sum =", total)
#input given:7
```

```python
# Find factorial of a number

n = int(input("Enter a number: "))
fact = 1

for i in range(1, n + 1):
    fact *= i

print("Factorial =", fact)
#input given:5
```

```python
# Count vowels in a string

text = input("Enter a string: ")
count = 0

for ch in text:
    if ch.lower() in "aeiou":
        count += 1

print("Number of vowels =", count)
#input given: "Hello World"
```
