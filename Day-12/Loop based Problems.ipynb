```python
#max element in list
list1 = [10, 20, 30, 40, 50]
max=float('-inf')
for i in list1:
    if i>max:
        max=i
print("Max element in list is:",max)
```

```python
#finding second largest element in list
list1 = [-2,-77,74,0,7,18]
first=second=float('-inf')
for i in list1:
    if i>first:
        second=first
        first=i
    elif i>second and i<first:
        second=i
print("Second largest element in list is:",second)
```

```python
#prime number check
n = int(input("Enter a number: "))
count=0
for i in range(1,n+1):
    if n%i==0:
        count+=1
if count==2:
    print(n,"is a prime number")
else:
    print(n,"is not a prime number")
```

```python
#reversing list elements
list1 = [1, 2, 3, 4, 5]
rev=[]
for num in list1:
    rev=[num]+rev
print("Reversed list is:",rev)
```
