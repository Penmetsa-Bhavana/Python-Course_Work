# EMAIL SENDING USING PYTHON

Python provides modules such as smtplib and email to send emails
programmatically.

## TYPES OF EMAIL SENDING

### 1. Normal Email
A simple email containing a message can be sent using Python.

### 2. Email with To, From and Subject
An email can include the sender's address, receiver's address,
subject, and message body.

### 3. Bulk Email
Bulk email is used to send the same email to multiple recipients.

### 4. Email with Attachment
An attachment such as a file, image, or document can be included
with an email.

### 5. Bulk Email with Attachment
The same email with an attachment can be sent to multiple recipients.

## IMPORTANT MODULES

### smtplib
Used to establish a connection with an SMTP server and send emails.

### email
Used to create email messages, subjects, attachments, and other
email components.

```python
import smtplib

sender = "sender@gmail.com"
receiver = "receiver@gmail.com"
password = "your_app_password"

message = "Hello, this is a normal email sent using Python."

server = smtplib.SMTP("smtp.gmail.com", 587)
server.starttls()
server.login(sender, password)

server.sendmail(sender, receiver, message)

server.quit()

print("Email sent successfully.")
```

```python
import smtplib
from email.message import EmailMessage

sender = "sender@gmail.com"
receiver = "receiver@gmail.com"
password = "your_app_password"

message = EmailMessage()

message["From"] = sender
message["To"] = receiver
message["Subject"] = "Python Email"

message.set_content("Hello, this email contains To, From and Subject.")

server = smtplib.SMTP("smtp.gmail.com", 587)
server.starttls()
server.login(sender, password)

server.send_message(message)

server.quit()

print("Email sent successfully.")
```

```python
import smtplib
from email.message import EmailMessage

sender = "sender@gmail.com"
password = "your_app_password"

receivers = [
    "receiver1@gmail.com",
    "receiver2@gmail.com",
    "receiver3@gmail.com"
]

server = smtplib.SMTP("smtp.gmail.com", 587)
server.starttls()
server.login(sender, password)

for receiver in receivers:
    message = EmailMessage()

    message["From"] = sender
    message["To"] = receiver
    message["Subject"] = "Python Bulk Email"

    message.set_content("Hello, this is a bulk email sent using Python.")

    server.send_message(message)

server.quit()

print("Bulk emails sent successfully.")
```

```python
import smtplib
from email.message import EmailMessage

sender = "sender@gmail.com"
receiver = "receiver@gmail.com"
password = "your_app_password"

message = EmailMessage()

message["From"] = sender
message["To"] = receiver
message["Subject"] = "Email with Attachment"

message.set_content("Hello, please find the attached file.")

file_path = "sample.txt"

with open(file_path, "rb") as file:
    file_data = file.read()

message.add_attachment(
    file_data,
    maintype="text",
    subtype="plain",
    filename="sample.txt"
)

server = smtplib.SMTP("smtp.gmail.com", 587)
server.starttls()
server.login(sender, password)

server.send_message(message)

server.quit()

print("Email with attachment sent successfully.")
```

```python
import smtplib
from email.message import EmailMessage

sender = "sender@gmail.com"
password = "your_app_password"

receivers = [
    "receiver1@gmail.com",
    "receiver2@gmail.com",
    "receiver3@gmail.com"
]

file_path = "sample.txt"

with open(file_path, "rb") as file:
    file_data = file.read()

server = smtplib.SMTP("smtp.gmail.com", 587)
server.starttls()
server.login(sender, password)

for receiver in receivers:
    message = EmailMessage()

    message["From"] = sender
    message["To"] = receiver
    message["Subject"] = "Bulk Email with Attachment"

    message.set_content("Hello, please find the attached file.")

    message.add_attachment(
        file_data,
        maintype="text",
        subtype="plain",
        filename="sample.txt"
    )

    server.send_message(message)

server.quit()

print("Bulk emails with attachment sent successfully.")
```
