CONDITIONAL STATEMENTS

## CONDITIONAL STATEMENTS

Conditional statements are used to make decisions in a program based on whether a condition is `True` or `False`.

Python provides four main types of conditional statements:
- `if`
- `if-else`
- `if-elif-else`
- Nested `if`

## IF STATEMENT

The `if` statement executes a block of code only when the given condition is `True`.

### SYNTAX

if condition:
    statement

### IMPORTANT POINTS
- The condition must return `True` or `False`.
- The statement inside `if` must be indented.
- If the condition is `False`, the `if` block is skipped.

## IF-ELSE STATEMENT

The `if-else` statement is used when there are two possible outcomes.

### SYNTAX

if condition:
    statement_if_true
else:
    statement_if_false

### IMPORTANT POINTS
- `if` block executes when the condition is `True`.
- `else` block executes when the condition is `False`.
- Only one of the two blocks is executed.

## IF-ELIF-ELSE STATEMENT

The `if-elif-else` statement is used when there are multiple conditions to check.

### SYNTAX

if condition1:
    statement1
elif condition2:
    statement2
elif condition3:
    statement3
else:
    statement4

### IMPORTANT POINTS
- Conditions are checked from top to bottom.
- The first `True` condition is executed.
- `elif` can be used multiple times.
- `else` is optional and executes when all conditions are `False`.

## NESTED IF STATEMENT

A nested `if` means placing one `if` statement inside another `if` or `else` block.

### SYNTAX

if condition1:
    if condition2:
        statement1
    else:
        statement2
else:
    statement3

### IMPORTANT POINTS
- Used when a second condition needs to be checked after the first condition is satisfied.
- Proper indentation is important.
- Multiple levels of nesting are possible.

## COMMON OPERATORS USED IN CONDITIONS

- Comparison: >, <, >=, <=, ==, !=
- Logical: and, or, not
- Membership: in, not in

```python
# Check whether a number is positive

num = int(input("Enter a number: "))

if num > 0:
    print("Positive number")
```

```python
# Check whether a number is even or odd

num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

```python
# Find grade based on marks

marks = int(input("Enter marks: "))

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
elif marks >= 40:
    print("Grade D")
else:
    print("Fail")
```

```python
# Check whether a number is positive and even or odd

num = int(input("Enter a number: "))

if num > 0:
    if num % 2 == 0:
        print("Positive even number")
    else:
        print("Positive odd number")
else:
    print("Number is not positive")
```
