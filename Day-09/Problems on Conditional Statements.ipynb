```python
#Fizz Buzz Program
num=int(input())
if num%3==0 and num%5==0:
    print("FizzBuzz")
elif num%3==0:
    print("Fizz")
elif num%5==0:
    print("Buzz")
```

```python
#Greatest among three numbers
a,b,c=map(int,input().split())
if a==b==c:
    print(a)
elif a==b:
    if c>a:
        print(c)
    else:
        print(a)
elif b==c:
    if a>b:
        print(a)
    else:
        print(b)
elif a==c:
    if b>a:
        print(b)
    else:
        print(a)
elif a>b and a>c:
    print(a)
elif b>a and b>c:
    print(b)
else:
    print(c)
```

```python
n=int(input())
if n>0:
    if n%2==0:
        print(f"{n} is a positive even number")
    else:
        print(f"{n} is a positive odd number")
elif n<0:
    if n%2==0:
        print(f"{n} is a negative even number")
    else:
        print(f"{n} is a negative odd number")
else:
    print(f"{n} is equal to zero")
```

```python
balance=int(input())
withdraw=int(input())
if withdraw<=balance:
    balance-=withdraw
    print(f"Withdrawn amount: {withdraw}")
    print(f"Remaining balance: {balance}")
else:
    print("Insufficient funds")
```

```python
amount=int(input())
if amount>=2000:
    discount=amount*(20/100)
    total_amount=amount-discount
    print(f"Total amount after discount: {total_amount}")
else:
    print(f"No Discount applied. Total amount: {amount}")
```
