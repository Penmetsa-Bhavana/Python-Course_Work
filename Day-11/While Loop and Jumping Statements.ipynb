# WHILE LOOP AND JUMPING STATEMENTS

## WHILE LOOP

A `while` loop repeatedly executes a block of code as long as the given condition is `True`.

## SYNTAX

while condition:
    statement

## IMPORTANT POINTS
- The condition is checked before every iteration.
- The loop stops when the condition becomes `False`.
- The loop variable must be updated inside the loop to avoid an infinite loop.

## BASIC PROBLEMS
- Print numbers from 1 to 10
- Print even numbers from 1 to 20
- Find the sum of numbers from 1 to n
- Find the factorial of a number
- Reverse a number
- Count the digits of a number

## JUMPING STATEMENTS

Jumping statements are used to change the normal flow of loop execution.

### BREAK
- Terminates the loop immediately.
- Control moves to the statement after the loop.

### CONTINUE
- Skips the current iteration.
- Control moves to the next iteration.

### PASS
- Does nothing.
- Used as a placeholder where a statement is syntactically required.

```python
# Print numbers from 1 to 10

i = 1

while i <= 10:
    print(i,end=" ")
    i += 1
```

```python
# Find factorial of a number

n = int(input("Enter a number: "))
fact = 1
i = 1

while i <= n:
    fact *= i
    i += 1

print("Factorial =", fact)
#given input: 5
```

```python
# Reverse a number

num = int(input("Enter a number: "))
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10

print("Reverse =", reverse)
#given input: 3456
```

```python
# Stop the loop when the number is 5

i = 1

while i <= 10:
    if i == 5:
        break
    print(i)
    i += 1
```

```python
# Skip number 5

i = 1

while i <= 10:
    if i == 5:
        i += 1
        continue
    print(i,end=" ")
    i += 1
```

```python
# pass acts as a placeholder

for i in range(1, 6):
    if i == 3:
        pass
    print(i)
```
