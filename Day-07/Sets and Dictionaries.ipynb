SETS AND DICTIONARIES

## SETS
- Unordered collection
- Mutable
- No duplicate values
- Different data types
- No indexing and slicing

## SET OPERATIONS
- Union (|)
- Intersection (&)
- Difference (-)
- Symmetric Difference (^)
- Membership (in, not in)

## SET METHODS
- add()
- update()
- remove()
- discard()
- pop()
- clear()
- copy()
- union()
- intersection()
- difference()
- symmetric_difference()

## FROZEN SET
- Immutable version of a set
- No modification methods
- Supports set operations
- Syntax: `frozenset(iterable)`

## DICTIONARY
- Collection of key-value pairs
- Mutable
- Ordered
- Keys must be unique
- Keys must be immutable/hashable

## DICTIONARY MAPPING
- Key → Value
- Access values using keys
- Add, update and delete key-value pairs

## DICTIONARY METHODS
- keys()
- values()
- items()
- get()
- update()
- pop()
- popitem()
- setdefault()
- clear()
- copy()

```python
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}

print(s1)
print(2 in s1)

print(s1 | s2)   # Union
print(s1 & s2)   # Intersection
print(s1 - s2)   # Difference
print(s1 ^ s2)   # Symmetric difference
```

```python
s = {1, 2, 3}

s.add(4)
print(s)

s.update([5, 6])
print(s)

s.remove(6)
print(s)

s.discard(10)    # No error if element is absent
print(s)

print(s.pop())
print(s)

copy_s = s.copy()
print(copy_s)

s.clear()
print(s)
```

```python
# Creating a frozen set
fs = frozenset([1, 2, 3, 4])

print(fs)

# Set operations are possible
fs2 = frozenset([3, 4, 5, 6])

print(fs | fs2)
print(fs & fs2)
print(fs - fs2)

# fs.add(5)      # Error - frozenset is immutable
```

```python
student = {
    "name": "Bhavana",
    "age": 21,
    "course": "Python"
}

print(student)

# Accessing values using keys
print(student["name"])
print(student.get("age"))

# Adding and updating
student["city"] = "Hyderabad"
student["age"] = 22

print(student)
```

```python
student = {
    "name": "Bhavana",
    "age": 21,
    "course": "Python"
}

print(student.keys())
print(student.values())
print(student.items())

student.update({"age": 22, "city": "Hyderabad"})
print(student)

print(student.pop("age"))
print(student)

print(student.popitem())
print(student)

student.setdefault("phone", "12345")
print(student)

copy_student = student.copy()
print(copy_student)

student.clear()
print(student)
```
