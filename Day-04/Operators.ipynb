OPERATORS IN PYTHON

TYPES:

1.Arithmetical operators: +,-,*,%,//,/,**
2.Assignment operators: =,+=,-=,*=,%=//=
3.Comparisional or Relational operators: ==,!=,>,<,>=,<=
4.Logical operators: and,or,not
5.Bitwise operators: &,|,^,<<,>>
6.Membership operators: in,not in
7.Identity operators: is,is not

```python
#Arithmetical Operators
a=20
b=10
print(a+b)
print(a-b)
print(a*b)
print(a%b)
print(a/b)
print(a//b)
```

```python
#Assignment Operators
a=10
a+=5
print(a)
a-=5
print(a)
a*=5
print(a)
a/=5
print(a)
```

```python
#Comparison Operators
a=20
b=10
print(a>b)
print(a<b)
print(a==b)
print(a!=b)
print(a>=b)
print(a<=b)
```

```python
#Logical Operators
a=True
b=False
print(a and b)
print(a or b)
print(not a)
```

```python
#Bitwise Operators
a=10
b=5
print(a & b)
print(a | b)
print(a ^ b)
print(~a)
print(a << 2)
print(a >> 2)
```

```python
#Membership Operators
a=10
b=[1,2,3,4,5]
print(a in b)
print(a not in b)
```

```python
#Identity Operators
a=10
b=10
print(a is b)
print(a is not b)
```
