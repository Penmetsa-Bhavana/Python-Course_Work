# NUMPY

NumPy stands for Numerical Python. It is a Python library used for
numerical computations and working with arrays.

It provides fast and efficient operations on large amounts of data.

## INSTALLING NUMPY

pip install numpy

## IMPORTING NUMPY

import numpy as np

## ARRAYS IN NUMPY

A NumPy array is a collection of elements of the same data type.
It supports one-dimensional, two-dimensional, and multidimensional arrays.

## CREATING ARRAYS

### 1. np.array()
Creates a NumPy array from a list or tuple.

### 2. np.zeros()
Creates an array filled with zeros.

### 3. np.ones()
Creates an array filled with ones.

### 4. np.arange()
Creates an array containing values within a specified range.

### 5. np.linspace()
Creates evenly spaced values within a specified interval.

## ARRAY ATTRIBUTES

### 1. ndim
Returns the number of dimensions.

### 2. shape
Returns the dimensions of an array.

### 3. size
Returns the total number of elements.

### 4. dtype
Returns the data type of the elements.

## ARRAY OPERATIONS

NumPy supports arithmetic operations such as addition, subtraction,
multiplication, and division on arrays.

## INDEXING AND SLICING

Indexing is used to access individual elements.
Slicing is used to access a range of elements.

## MATHEMATICAL FUNCTIONS

NumPy provides functions such as sum(), min(), max(), mean(),
sqrt(), and sort().

## RESHAPING ARRAYS

The reshape() function changes the shape of an array without
changing its data, provided the total number of elements remains the same.

## ARRAY JOINING AND SPLITTING

The concatenate() function joins arrays.
The split() function divides an array into smaller arrays.

```python
# Creating a NumPy array

import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

print(numbers)
print(type(numbers))
```

```python
# Creating arrays using NumPy functions

import numpy as np

print(np.zeros(5))
print(np.ones(5))
print(np.arange(1, 11))
print(np.linspace(1, 10, 5))
```

```python
# Creating a two-dimensional array

import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])

print(arr)
```

```python
# NumPy array attributes

import numpy as np

arr = np.array([[10, 20, 30], [40, 50, 60]])

print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Data type:", arr.dtype)
```

```python
# Indexing and slicing

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("First element:", arr[0])
print("Last element:", arr[-1])
print("First three elements:", arr[:3])
print("Elements from index 2:", arr[2:])
```

```python
# Arithmetic operations on arrays

import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
# Mathematical functions

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))
print("Mean:", np.mean(arr))
print("Square root:", np.sqrt(arr))
print("Sorted array:", np.sort(arr))
```

```python
# Reshaping an array

import numpy as np

arr = np.arange(1, 10)

new_arr = arr.reshape(3, 3)

print(new_arr)
```

```python
# Joining arrays

import numpy as np

a = np.array([10, 20, 30])
b = np.array([40, 50, 60])

result = np.concatenate((a, b))

print(result)
```

```python
# Splitting an array

import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60])

result = np.split(arr, 3)

print(result)
```

```python
# Filtering array elements

import numpy as np

arr = np.array([10, 25, 30, 45, 50, 65])

result = arr[arr > 30]

print(result)
```
